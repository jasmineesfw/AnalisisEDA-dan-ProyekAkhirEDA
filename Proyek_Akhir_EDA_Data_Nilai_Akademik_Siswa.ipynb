```python
import numpy as np
import pandas as pd

df = pd.read_csv("dataset_nilai_akademik_siswa.csv")
```

```python
# Menampilkan 5 baris pertama
print("=== 5 Baris Pertama Dataset ===")
display(df.head())

# Menampilkan struktur data
print("\n=== Informasi Struktur & Tipe Data ===")
df.info()

# Menampilan statistik data
print("\n=== Statistik Ringkas Kolom Numerik ===")
display(df.describe())

# Menampilkan dimensi dataset (baris, kolom)
print(f"\nJumlah total baris dan kolom: {df.shape}")
```

```python
# Cek Missing Value
print("=== Jumlah Missing Value Sebelum Cleaning ===")
print(df.isnull().sum())

# Cek Duplikat
jumlah_duplikat = df.duplicated().sum()
print(f"\nJumlah baris duplikat: {jumlah_duplikat}")

# Hapus duplikat jika ditemukan
if jumlah_duplikat > 0:
    df = df.drop_duplicates()
    print("Duplikat berhasil dihapus.")
else:
    print("Tidak ada data duplikat.")

# Membersihkan kolom nilai
df['nilai_clean'] = (
    df['nilai']
    .astype(str)
    .str.replace(',', '.', regex=False)
    .str.replace(' poin', '', regex=False)
    .str.strip()
)

# Mengubah nilai menjadi numerik
df['nilai_clean'] = pd.to_numeric(
    df['nilai_clean'],
    errors='coerce'
)

# Mengubah nilai 999 menjadi missing
df['nilai_clean'] = df['nilai_clean'].replace(999, np.nan)

# Mengisi missing value nilai dengan median
median_nilai = df['nilai_clean'].median()
df['nilai_clean'] = df['nilai_clean'].fillna(median_nilai)

# Menimpa kolom nilai asli dengan nilai_clean
df['nilai'] = df['nilai_clean']

# Mengisi missing value guru_pengampu
df['guru_pengampu'] = df['guru_pengampu'].fillna(
    'Tidak Diketahui'
)

# Membersihkan tanggal ujian
bulan = {
    'Januari': 'January',
    'Februari': 'February',
    'Maret': 'March',
    'April': 'April',
    'Mei': 'May',
    'Juni': 'June',
    'Juli': 'July',
    'Agustus': 'August',
    'September': 'September',
    'Oktober': 'October',
    'November': 'November',
    'Desember': 'December'
}
for indo, eng in bulan.items():
    df['tanggal_ujian'] = df['tanggal_ujian'].str.replace(
        indo,
        eng,
        regex=False
    )

# Mengubah tanggal menjadi datetime
df['tanggal_ujian'] = pd.to_datetime(
    df['tanggal_ujian'],
    dayfirst=True,
    errors='coerce'
)

# Mengisi missing value pada tanggal_ujian (agar tidak NaT)
df['tanggal_ujian'] = df['tanggal_ujian'].fillna(pd.Timestamp('2026-08-01'))

# Cek Missing Value setelah cleaning
print("\n=== Missing Value Setelah Cleaning ===")
print(df.isnull().sum())

# Cek tipe data setelah cleaning
print("\n=== Tipe Data Setelah Cleaning ===")
print(df.dtypes)
```

```python
# Kolom Turunan: Status Kelulusan
df['Status'] = np.where(
    df['nilai_clean'] >= 75,
    'Lulus',
    'Remedial'
)
print("=== Data dengan Status Kelulusan ===")
display(
    df[['nama', 'kelas', 'mata_pelajaran', 'nilai_clean', 'Status']].head()
)

# Filtering: Siswa yang perlu remedial
siswa_remedial = df[df['Status'] == 'Remedial']
print("\n=== Daftar Siswa Remedial ===")
display(
    siswa_remedial[
        ['nama', 'kelas', 'mata_pelajaran', 'nilai_clean', 'Status']
    ]
)

# Sorting: Nilai tertinggi ke terendah
df_sorted = df.sort_values(
    by='nilai_clean',
    ascending=False
)
print("\n=== Top 5 Nilai Tertinggi ===")
display(
    df_sorted[
        ['nama', 'kelas', 'mata_pelajaran', 'nilai_clean']
    ].head()
)

# Groupby/agregasi: Statistik nilai per kelas
ringkasan_nilai = df.groupby('kelas')['nilai_clean'].agg(
    ['mean', 'min', 'max', 'count']
).round(2)
print("\n=== Agregasi Nilai Per Kelas ===")
display(ringkasan_nilai)

# Groupby/agregasi: Rata-rata nilai per mata pelajaran
ringkasan_mapel = df.groupby(
    'mata_pelajaran'
)['nilai_clean'].agg(
    ['mean', 'min', 'max', 'count']
).round(2)
print("\n=== Agregasi Nilai Per Mata Pelajaran ===")
display(ringkasan_mapel)
```

```python
# Simpan dataset yang sudah bersih ke file CSV baru
df.to_csv("dataset_nilai_akademik_bersih.csv", index=False)
print("Dataset bersih berhasil disimpan sebagai 'dataset_nilai_akademik_bersih.csv'!")
```

```python
print("=== Cek Akhir Dataset Bersih ===")
print("\nMissing Value:")
print(df.isnull().sum())

print("\nJumlah Duplikat:")
print(df.duplicated().sum())

print("\nTipe Data:")
print(df.dtypes)
```

```python
# Data Profiling Summary
rata_rata = df['nilai_clean'].mean()
nilai_tertinggi = df['nilai_clean'].max()
nilai_terendah = df['nilai_clean'].min()

jumlah_remedial = (df['Status'] == 'Remedial').sum()
jumlah_siswa = len(df)

kelas_tertinggi = (
    df.groupby('kelas')['nilai_clean']
    .mean()
    .idxmax()
)

print("=== Data Profiling Summary ===")
print(
    f"1. Rata-rata nilai keseluruhan siswa adalah "
    f"{rata_rata:.2f}, dengan nilai tertinggi "
    f"{nilai_tertinggi:.2f} dan nilai terendah "
    f"{nilai_terendah:.2f}."
)

print(
    f"2. Terdapat {jumlah_remedial} dari {jumlah_siswa} "
    f"siswa yang mendapatkan status Remedial "
    f"karena nilainya di bawah KKM 75."
)

print(
    f"3. Berdasarkan rata-rata nilai per kelas, "
    f"kelas {kelas_tertinggi} memiliki rata-rata "
    f"nilai tertinggi."
)
```

2. Terdapat beberapa siswa dengan status Remedial karena nilai berada di bawah KKM 75. Data ini dapat digunakan untuk mengetahui siswa yang membutuhkan perhatian atau pembelajaran tambahan.
