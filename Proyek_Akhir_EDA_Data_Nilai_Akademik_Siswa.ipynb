```python
import numpy as np
import pandas as pd

df = pd.read_csv("dataset_nilai_akademik_siswa.csv")
```

```python
# Menampilkan 5 baris pertama
print("=== 5 Baris Pertama Dataset ===")
display(df.head())

# Menampilkan struktur data
print("\n=== Informasi Struktur & Tipe Data ===")
df.info()

# Menampilan statistik data
print("\n=== Statistik Ringkas Kolom Numerik ===")
display(df.describe())

# Menampilkan dimensi dataset (baris, kolom)
print(f"\nJumlah total baris dan kolom: {df.shape}")
```

```python
# Cek & Tangani Missing Value
print("Jumlah data kosong per kolom:")
print(df.isnull().sum())

# Cek & Tangani Data Duplikat
jumlah_duplikat = df.duplicated().sum()
print(f"\nJumlah baris duplikat: {jumlah_duplikat}")
if jumlah_duplikat > 0:
  df = df.drop_duplicates()
  print("Duplikat berhasil dihapus.")

print("\n=== Tipe Data Setelah Pembersihan ===")
print(df.dtypes)
```

```python
# Bersihkan kolom nilai agar menjadi angka murni (float)
df['nilai_clean'] = (
    df['nilai']
    .astype(str)
    .str.replace(',', '.')  # Ubah koma menjadi titik
    .str.replace(' poin', '')  # Hapus kata 'poin'
    .str.strip()
)

# Konversi ke angka, ganti data yang tidak valid/kosong dengan NaN
df['nilai_clean'] = pd.to_numeric(df['nilai_clean'], errors='coerce')

# Tangani nilai outlier 999 (ganti dengan NaN lalu isi dengan median/mean)
df['nilai_clean'] = df['nilai_clean'].replace(999, np.nan)
df['nilai_clean'] = df['nilai_clean'].fillna(df['nilai_clean'].median())

# Kolom Turunan: Menentukan status kelulusan (KKM = 75)
df['Status'] = np.where(df['nilai_clean'] >= 75, 'Lulus', 'Remedial')

# Filtering: Filter siswa yang perlu remedial
siswa_remedial = df[df['Status'] == 'Remedial']
print("=== Daftar Siswa Remedial ===")
display(siswa_remedial[['nama', 'kelas', 'mata_pelajaran', 'nilai_clean', 'Status']].head())

# Sorting: Mengurutkan nilai siswa dari yang tertinggi
df_sorted = df.sort_values(by='nilai_clean', ascending=False)
print("\n=== Top 5 Nilai Tertinggi ===")
display(df_sorted[['nama', 'kelas', 'mata_pelajaran', 'nilai_clean']].head())

# Groupby/agregasi: Menghitung statistik nilai per kelas
ringkasan_nilai = df.groupby('kelas')['nilai_clean'].agg(
    ['mean', 'min', 'max', 'count']
).round(2)

print("\n=== Agregasi Nilai Per Kelas ===")
display(ringkasan_nilai)
```

```python
# Simpan dataset yang sudah bersih ke file CSV baru
df.to_csv("dataset_nilai_akademik_bersih.csv", index=False)
print("Dataset bersih berhasil disimpan sebagai 'dataset_nilai_akademik_bersih.csv'!")
```
