```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)

# INI ERROR
# Bandingkan hasil harga * 0.9 di atas dengan jika harga berupa list Python biasa (bukan array).
# harga = [5000, 7000, 3000]
# print(harga * 0.9)
```

**Tugas Analisis 1**

NumPy Array bisa melakukan operasi matematika langsung pada setiap elemennya (vectorized operation), sedangkan list Python biasa tidak bisa melakukan operasi tersebut secara langsung.

```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

**Tugas Analisis 2**

Kolom **menu** di index ke-4 dan kolom **terjual** di index ke-2. Jika data tersebut langsung dianalisis tanpa dibersihkan, akan menyebabkan data bias/hasil analisis kurang akurat.

```python
df = pd.read_csv('data_kantin.csv')

print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

**Tugas Analisis 3**

Kolom **terjual** memiliki jumlah non-null lebih sedikit dari total baris, jumlah null pada kolom **terjual** ada 6 data.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom

df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

**Tugas Analisis 4**

**terjual** diisi 0 karena data yang kosong dianggap tidak ada penjualan. Sedangkan **menu** yang kosong dihapus karena nama menu merupakan data penting dan bisa mengganggu hasil analisis jika diisi dengan fillna().


```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

**Tugas Analisis 5**

105 baris, karena ada 5 baris duplikat yang dihapus. Memastikan tipe data (dtypes), penting agar data bisa diolah dan dianalisis dengan benar sesuai jenis datanya.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```
